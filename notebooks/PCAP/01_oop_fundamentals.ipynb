## Classes and Objects

Understanding the fundamental building blocks of OOP in Python.

```python
# Basic class definition
class Student:
    """A simple Student class."""
    
    # Class variable (shared by all instances)
    school = "Python University"
    
    def __init__(self, name, age, grade):
        """Initialize a new student."""
        self.name = name      # Instance variable
        self.age = age        # Instance variable
        self.grade = grade    # Instance variable
    
    def introduce(self):
        """Student introduces themselves."""
        return f"Hi, I'm {self.name}, {self.age} years old, grade {self.grade}"
    
    def study(self, subject):
        """Student studies a subject."""
        return f"{self.name} is studying {subject}"

# Create instances
alice = Student("Alice", 20, "A")
bob = Student("Bob", 19, "B")

print(alice.introduce())
print(bob.study("Python Programming"))
print(f"School: {Student.school}")
```

## Inheritance

Creating classes that inherit from other classes.

```python
# Base class
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def greet(self):
        return f"Hello, I'm {self.name}"
    
    def get_info(self):
        return f"Name: {self.name}, Age: {self.age}"

# Derived class
class Teacher(Person):
    def __init__(self, name, age, subject):
        super().__init__(name, age)  # Call parent constructor
        self.subject = subject
    
    def teach(self):
        return f"Professor {self.name} is teaching {self.subject}"
    
    def get_info(self):  # Method overriding
        return super().get_info() + f", Subject: {self.subject}"

# Another derived class
class Student(Person):
    def __init__(self, name, age, student_id):
        super().__init__(name, age)
        self.student_id = student_id
    
    def study(self):
        return f"{self.name} (ID: {self.student_id}) is studying"
    
    def get_info(self):
        return super().get_info() + f", Student ID: {self.student_id}"

# Create instances
teacher = Teacher("Dr. Smith", 45, "Computer Science")
student = Student("Alice", 20, "S12345")

print(teacher.greet())
print(teacher.teach())
print(teacher.get_info())
print()
print(student.greet())
print(student.study())
print(student.get_info())
```

## Encapsulation and Properties

Protecting data and providing controlled access.

```python
class BankAccount:
    def __init__(self, account_number, initial_balance=0):
        self.account_number = account_number
        self._balance = initial_balance  # Protected attribute
        self.__pin = "1234"  # Private attribute
    
    @property
    def balance(self):
        """Get the current balance."""
        return self._balance
    
    @balance.setter
    def balance(self, amount):
        """Set balance with validation."""
        if amount >= 0:
            self._balance = amount
        else:
            raise ValueError("Balance cannot be negative")
    
    def deposit(self, amount):
        """Deposit money to account."""
        if amount > 0:
            self._balance += amount
            return f"Deposited ${amount}. New balance: ${self._balance}"
        else:
            return "Invalid deposit amount"
    
    def withdraw(self, amount):
        """Withdraw money from account."""
        if 0 < amount <= self._balance:
            self._balance -= amount
            return f"Withdrew ${amount}. New balance: ${self._balance}"
        else:
            return "Insufficient funds or invalid amount"
    
    def _validate_pin(self, pin):
        """Protected method to validate PIN."""
        return pin == self.__pin

# Create and use account
account = BankAccount("ACC123", 1000)

print(f"Initial balance: ${account.balance}")
print(account.deposit(500))
print(account.withdraw(200))
print(f"Final balance: ${account.balance}")

# Try to access private attribute (will show name mangling)
print(f"Private PIN access: {hasattr(account, '__pin')}")
print(f"Mangled name exists: {hasattr(account, '_BankAccount__pin')}")
```

## Polymorphism

Different objects responding to the same interface.

```python
from abc import ABC, abstractmethod

# Abstract base class
class Animal(ABC):
    def __init__(self, name):
        self.name = name
    
    @abstractmethod
    def make_sound(self):
        pass
    
    def eat(self):
        return f"{self.name} is eating"

# Concrete classes
class Dog(Animal):
    def make_sound(self):
        return f"{self.name} says Woof!"

class Cat(Animal):
    def make_sound(self):
        return f"{self.name} says Meow!"

class Bird(Animal):
    def make_sound(self):
        return f"{self.name} says Tweet!"
    
    def fly(self):
        return f"{self.name} is flying"

# Polymorphism in action
def animal_concert(animals):
    for animal in animals:
        print(animal.make_sound())
        print(animal.eat())
        if hasattr(animal, 'fly'):
            print(animal.fly())
        print()

# Create different animals
dog = Dog("Buddy")
cat = Cat("Whiskers")
bird = Bird("Tweety")

animals = [dog, cat, bird]
animal_concert(animals)
```

## Special Methods (Magic Methods)

Implementing Python's special methods for custom behavior.

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    
    def __str__(self):
        """String representation for users."""
        return f"Vector({self.x}, {self.y})"
    
    def __repr__(self):
        """String representation for developers."""
        return f"Vector(x={self.x}, y={self.y})"
    
    def __add__(self, other):
        """Vector addition."""
        if isinstance(other, Vector):
            return Vector(self.x + other.x, self.y + other.y)
        return NotImplemented
    
    def __sub__(self, other):
        """Vector subtraction."""
        if isinstance(other, Vector):
            return Vector(self.x - other.x, self.y - other.y)
        return NotImplemented
    
    def __mul__(self, scalar):
        """Scalar multiplication."""
        if isinstance(scalar, (int, float)):
            return Vector(self.x * scalar, self.y * scalar)
        return NotImplemented
    
    def __eq__(self, other):
        """Equality comparison."""
        if isinstance(other, Vector):
            return self.x == other.x and self.y == other.y
        return False
    
    def __len__(self):
        """Magnitude of vector."""
        return int((self.x**2 + self.y**2)**0.5)

# Test the Vector class
v1 = Vector(3, 4)
v2 = Vector(1, 2)

print(f"v1: {v1}")
print(f"v2: {v2}")
print(f"v1 + v2: {v1 + v2}")
print(f"v1 - v2: {v1 - v2}")
print(f"v1 * 2: {v1 * 2}")
print(f"v1 == v2: {v1 == v2}")
print(f"Length of v1: {len(v1)}")
print(f"Repr of v1: {repr(v1)}")
```

## Practice Exercise: Library Management System

Create a comprehensive library management system using OOP concepts.

```python
from datetime import datetime, timedelta

class Book:
    def __init__(self, title, author, isbn):
        self.title = title
        self.author = author
        self.isbn = isbn
        self.is_available = True
        self.due_date = None
    
    def __str__(self):
        status = "Available" if self.is_available else f"Due: {self.due_date}"
        return f"{self.title} by {self.author} - {status}"

class Member:
    def __init__(self, name, member_id):
        self.name = name
        self.member_id = member_id
        self.borrowed_books = []
    
    def borrow_book(self, book):
        if book.is_available:
            book.is_available = False
            book.due_date = datetime.now() + timedelta(days=14)
            self.borrowed_books.append(book)
            return f"{self.name} borrowed '{book.title}'"
        else:
            return f"'{book.title}' is not available"
    
    def return_book(self, book):
        if book in self.borrowed_books:
            book.is_available = True
            book.due_date = None
            self.borrowed_books.remove(book)
            return f"{self.name} returned '{book.title}'"
        else:
            return f"{self.name} doesn't have '{book.title}'"

class Library:
    def __init__(self, name):
        self.name = name
        self.books = []
        self.members = []
    
    def add_book(self, book):
        self.books.append(book)
        return f"Added '{book.title}' to {self.name}"
    
    def register_member(self, member):
        self.members.append(member)
        return f"Registered {member.name} as member"
    
    def find_book(self, title):
        for book in self.books:
            if book.title.lower() == title.lower():
                return book
        return None
    
    def available_books(self):
        return [book for book in self.books if book.is_available]

# Create library system
library = Library("Python Central Library")

# Add books
book1 = Book("Python Programming", "John Doe", "123456789")
book2 = Book("Data Structures", "Jane Smith", "987654321")
book3 = Book("Web Development", "Bob Johnson", "456789123")

print(library.add_book(book1))
print(library.add_book(book2))
print(library.add_book(book3))

# Register members
member1 = Member("Alice", "M001")
member2 = Member("Bob", "M002")

print(library.register_member(member1))
print(library.register_member(member2))

# Borrow and return books
print("\n--- Borrowing Books ---")
print(member1.borrow_book(book1))
print(member2.borrow_book(book2))
print(member1.borrow_book(book1))  # Try to borrow same book

print("\n--- Book Status ---")
for book in library.books:
    print(book)

print("\n--- Available Books ---")
for book in library.available_books():
    print(f"- {book.title}")

print("\n--- Returning Books ---")
print(member1.return_book(book1))
print(f"Available books after return: {len(library.available_books())}")
```
