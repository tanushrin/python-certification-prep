## Lists - Dynamic Arrays

Lists are mutable, ordered collections of items.

```python
# Creating lists
empty_list = []
numbers = [1, 2, 3, 4, 5]
mixed = [1, "hello", 3.14, True]

print(f"Empty list: {empty_list}")
print(f"Numbers: {numbers}")
print(f"Mixed types: {mixed}")
print(f"Length of numbers: {len(numbers)}")
```

```python
# List operations
fruits = ["apple", "banana", "orange"]

# Accessing elements
print(f"First fruit: {fruits[0]}")
print(f"Last fruit: {fruits[-1]}")

# Modifying lists
fruits.append("grape")
print(f"After append: {fruits}")

fruits.insert(1, "kiwi")
print(f"After insert: {fruits}")

removed = fruits.pop()
print(f"Removed: {removed}, List: {fruits}")
```

## Dictionaries - Key-Value Pairs

Dictionaries store data as key-value pairs.

```python
# Creating dictionaries
empty_dict = {}
student = {
    "name": "Alice",
    "age": 20,
    "grade": "A",
    "subjects": ["Math", "Physics", "Chemistry"]
}

print(f"Student info: {student}")
print(f"Student name: {student['name']}")
print(f"Student subjects: {student['subjects']}")
```

```python
# Dictionary operations
student["email"] = "alice@example.com"  # Add new key-value
student["age"] = 21  # Update existing value

print(f"Updated student: {student}")

# Dictionary methods
print(f"Keys: {list(student.keys())}")
print(f"Values: {list(student.values())}")
print(f"Items: {list(student.items())}")
```

## Tuples - Immutable Sequences

Tuples are immutable, ordered collections.

```python
# Creating tuples
coordinates = (10, 20)
colors = ("red", "green", "blue")
single_item = (42,)  # Note the comma for single-item tuple

print(f"Coordinates: {coordinates}")
print(f"Colors: {colors}")
print(f"Single item: {single_item}")

# Accessing tuple elements
x, y = coordinates  # Tuple unpacking
print(f"X: {x}, Y: {y}")
```

## Sets - Unique Collections

Sets store unique elements.

```python
# Creating sets
empty_set = set()
numbers = {1, 2, 3, 4, 5}
duplicates = {1, 2, 2, 3, 3, 4}  # Duplicates are automatically removed

print(f"Numbers set: {numbers}")
print(f"Set with duplicates removed: {duplicates}")

# Set operations
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}

print(f"Union: {set1 | set2}")
print(f"Intersection: {set1 & set2}")
print(f"Difference: {set1 - set2}")
```

## Practice Challenge: Student Management System

Create a simple student management system using the data structures you've learned.

```python
# Student management system
students = [
    {"name": "Alice", "age": 20, "grades": [85, 92, 78, 96]},
    {"name": "Bob", "age": 19, "grades": [78, 85, 82, 79]},
    {"name": "Charlie", "age": 21, "grades": [92, 88, 91, 87]}
]

# Calculate average grade for each student
for student in students:
    grades = student["grades"]
    average = sum(grades) / len(grades)
    student["average"] = round(average, 2)
    print(f"{student['name']}: {student['average']}%")

# Find the best student
best_student = max(students, key=lambda s: s["average"])
print(f"\nBest student: {best_student['name']} ({best_student['average']}%)")
```
